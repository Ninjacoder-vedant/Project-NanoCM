````python
import os
import re
import subprocess
import tempfile
from pathlib import Path
from functools import lru_cache
from concurrent.futures import ProcessPoolExecutor, as_completed

import pandas as pd

GENERATED_TESTS_DIR = Path("hf_cache/generated_tests")

# ---------- Load generated tests for a problem ----------
@lru_cache(maxsize=128)
def _load_contest_parquet(contest_id: int) -> pd.DataFrame | None:
    """Cache parquet files per contest (many problems share one file)."""
    path = GENERATED_TESTS_DIR / f"test_cases_{contest_id:04d}.parquet"
    if not path.exists():
        return None
    return pd.read_parquet(path)

def load_generated_tests(problem_id: str, contest_id: int) -> list[dict]:
    """Return list of {'input': ..., 'output': ...} for the given problem."""
    df = _load_contest_parquet(contest_id)
    if df is None:
        return []
    sub = df[df["problem_id"] == problem_id]
    return [{"input": r["input"], "output": r["output"]} for _, r in sub.iterrows()]


# ---------- Extract, compile, run, check (same as before) ----------
CODE_RE = re.compile(r"```cpp\s*\n(.*?)```", re.DOTALL | re.IGNORECASE)

def extract_cpp(response: str) -> str | None:
    matches = CODE_RE.findall(response)
    if not matches:
        m = re.search(r"```\s*\n(.*?)```", response, re.DOTALL)
        return m.group(1) if m else None
    return matches[-1]

def compile_cpp(code: str, workdir: Path) -> tuple[Path | None, str]:
    src = workdir / "sol.cpp"
    exe = workdir / "sol"
    src.write_text(code)
    proc = subprocess.run(
        ["g++", "-O2", "-std=c++17", "-o", str(exe), str(src)],
        capture_output=True, text=True, timeout=30,
    )
    if proc.returncode != 0:
        return None, proc.stderr
    return exe, ""

def run_binary(exe: Path, stdin_data: str, time_limit: float) -> tuple[str, str]:
    try:
        proc = subprocess.run(
            [str(exe)], input=stdin_data, capture_output=True, text=True,
            timeout=time_limit + 0.5,
        )
    except subprocess.TimeoutExpired:
        return "tle", ""
    if proc.returncode != 0:
        return "rte", proc.stdout
    return "ok", proc.stdout

def normalize(s: str) -> list[list[str]]:
    lines = [ln.rstrip() for ln in s.replace("\r\n", "\n").split("\n")]
    while lines and lines[-1] == "":
        lines.pop()
    return [ln.split() for ln in lines]

def direct_match(expected: str, actual: str) -> bool:
    return normalize(expected) == normalize(actual)

def checker_match(checker_src: str, inp: str, ref_out: str, sub_out: str,
                  workdir: Path) -> bool:
    chk = workdir / "checker.py"; chk.write_text(checker_src)
    inp_f = workdir / "in.txt";   inp_f.write_text(inp)
    ref_f = workdir / "ref.txt";  ref_f.write_text(ref_out)
    sub_f = workdir / "sub.txt";  sub_f.write_text(sub_out)
    try:
        proc = subprocess.run(
            ["python3", str(chk), str(inp_f), str(ref_f), str(sub_f)],
            capture_output=True, text=True, timeout=30,
        )
    except subprocess.TimeoutExpired:
        return False
    return proc.stdout.strip() == "1"


# ---------- Run one set of tests against a compiled binary ----------
def run_tests(exe: Path, tests: list[dict], time_limit: float,
              checker_src: str | None, workdir: Path) -> dict:
    passed, total = 0, len(tests)
    per_test = []
    for t in tests:
        status, out = run_binary(exe, t["input"], time_limit)
        if status == "tle":
            per_test.append("tle"); continue
        if status == "rte":
            per_test.append("rte"); continue
        if checker_src:
            ok = checker_match(checker_src, t["input"], t["output"], out, workdir)
        else:
            ok = direct_match(t["output"], out)
        if ok:
            passed += 1
            per_test.append("ok")
        else:
            per_test.append("wa")
    return {"passed": passed, "total": total, "per_test": per_test}


# ---------- Evaluate one problem on official + generated ----------
def evaluate_problem(row: dict, llm_response: str) -> dict:
    result = {
        "id": row["id"],
        "compile_ok": False, "error": None,
        "official_passed": 0, "official_total": 0, "official_accuracy": 0.0,
        "generated_passed": 0, "generated_total": 0, "generated_accuracy": 0.0,
        "total_passed": 0, "total_total": 0, "total_accuracy": 0.0,
        "official_per_test": [], "generated_per_test": [],
    }

    code = extract_cpp(llm_response)
    if code is None:
        result["error"] = "no_code_extracted"
        return result

    # Official tests
    official = _coerce_tests(row.get("official_tests"))
    print(type(row["official_tests"]))

    # Generated tests
    generated = load_generated_tests(row["id"], int(row["contest_id"]))

    if not official and not generated:
        result["error"] = "no_tests"
        return result

    time_limit = float(row.get("time_limit") or 1.0)
    checker_src = row.get("generated_checker")

    with tempfile.TemporaryDirectory() as td:
        workdir = Path(td)
        exe, err = compile_cpp(code, workdir)
        if exe is None:
            result["error"] = f"compile_error: {err[:500]}"
            return result
        result["compile_ok"] = True

        if official:
            r = run_tests(exe, official, time_limit, checker_src, workdir)
            result["official_passed"] = r["passed"]
            result["official_total"]  = r["total"]
            result["official_per_test"] = r["per_test"]
            result["official_accuracy"] = r["passed"] / r["total"] if r["total"] else 0.0

        if generated:
            r = run_tests(exe, generated, time_limit, checker_src, workdir)
            result["generated_passed"] = r["passed"]
            result["generated_total"]  = r["total"]
            result["generated_per_test"] = r["per_test"]
            result["generated_accuracy"] = r["passed"] / r["total"] if r["total"] else 0.0

    result["total_passed"] = result["official_passed"] + result["generated_passed"]
    result["total_total"]  = result["official_total"]  + result["generated_total"]
    if result["total_total"]:
        result["total_accuracy"] = result["total_passed"] / result["total_total"]
    return result


# ---------- Parallel driver ----------
def _worker(args):
    row_dict, llm_response = args
    return evaluate_problem(row_dict, llm_response)

def evaluate_dataframe(df, response_col="llm_response", n_workers=1):
    tasks = [(row, row[response_col]) for row in df.to_dict("records")]
    results = [None] * len(tasks)
    with ProcessPoolExecutor(max_workers=n_workers) as ex:
        futs = {ex.submit(_worker, t): i for i, t in enumerate(tasks)}
        for fut in as_completed(futs):
            results[futs[fut]] = fut.result()

    df = df.copy()
    for key in ["compile_ok", "error",
                "official_passed", "official_total", "official_accuracy",
                "generated_passed", "generated_total", "generated_accuracy",
                "total_passed", "total_total", "total_accuracy",
                "official_per_test", "generated_per_test"]:
        df[key] = [r[key] for r in results]
    return df
````

````python
import json
from pathlib import Path
from tqdm import tqdm
import torch


def _build_chat_text(tokenizer, prompt: str) -> str:
    messages = [
        {"role": "system", "content": "Do all the reasoning before giving code output in <think>..</think> tags."},
        {"role": "user", "content": prompt},
    ]
    return tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )


def generate_batch(model, tokenizer, prompts: list[str], max_new_tokens: int = 16000) -> list[str]:
    texts = [_build_chat_text(tokenizer, p) for p in prompts]

    tokenizer.padding_side = "left"
    if tokenizer.pad_token is None:
        tokenizer.pad_token = tokenizer.eos_token

    inputs = tokenizer(
        texts, return_tensors="pt", padding=True, truncation=False,
    ).to(model.device)

    input_len = inputs["input_ids"].shape[1]

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=0.7,
            top_p=0.8,
            top_k=20,
            pad_token_id=tokenizer.pad_token_id,
        )

    new_tokens = outputs[:, input_len:]
    return tokenizer.batch_decode(new_tokens, skip_special_tokens=True)


def _safe_filename(pid: str) -> str:
    """Convert '1993/A' -> '1993_A' so it's a valid filename."""
    return pid.replace("/", "_").replace("\\", "_")


def run_inference(
    model,
    tokenizer,
    df,
    output_path: str | Path = "responses.json",
    per_problem_dir: str | Path = "responses",
    batch_size: int = 4,
    max_new_tokens: int = 16000,
    id_col: str = "id",
    prompt_col: str = "prompt",
    resume: bool = True,
):
    output_path = Path(output_path)
    per_problem_dir = Path(per_problem_dir)
    per_problem_dir.mkdir(parents=True, exist_ok=True)

    # Load existing results if resuming
    results: dict[str, str] = {}
    if resume and output_path.exists():
        with open(output_path) as f:
            results = json.load(f)
        print(f"Resuming — {len(results)} problems already done.")

    todo = [(row[id_col], row[prompt_col])
            for _, row in df.iterrows()
            if row[id_col] not in results]

    if not todo:
        print("Nothing to do.")
        return results

    for start in tqdm(range(0, len(todo), batch_size), desc="Batches"):
        batch = todo[start : start + batch_size]
        ids = [b[0] for b in batch]
        prompts = [b[1] for b in batch]

        try:
            responses = generate_batch(model, tokenizer, prompts, max_new_tokens)
        except torch.cuda.OutOfMemoryError:
            print(f"OOM at batch starting {start}. Try a smaller batch_size.")
            torch.cuda.empty_cache()
            raise

        for pid, prompt, resp in zip(ids, prompts, responses):
            results[pid] = resp

            # Save each response individually for easy inspection
            per_file = per_problem_dir / f"{_safe_filename(pid)}.json"
            with open(per_file, "w", encoding="utf-8") as f:
                json.dump(
                    {"id": pid, "prompt": prompt, "response": resp},
                    f, ensure_ascii=False, indent=2,
                )

        # Save aggregate after every batch
        with open(output_path, "w", encoding="utf-8") as f:
            json.dump(results, f, ensure_ascii=False, indent=2)

    print(f"Done. {len(results)} responses → {output_path}, per-problem → {per_problem_dir}/")
    return results
````

````python
import os
import re
import subprocess
import tempfile
import shutil
import ast
from pathlib import Path
from concurrent.futures import ProcessPoolExecutor, as_completed


# ---------- 1. Extract C++ code from LLM response ----------
CODE_RE = re.compile(r"```cpp\s*\n(.*?)```", re.DOTALL | re.IGNORECASE)

def extract_cpp(response: str) -> str | None:
    matches = CODE_RE.findall(response)
    if not matches:
        # fallback: any ``` block
        m = re.search(r"```\s*\n(.*?)```", response, re.DOTALL)
        return m.group(1) if m else None
    return matches[-1]  # take last code block (final answer)


# ---------- 2. Compile ----------
def compile_cpp(code: str, workdir: Path) -> tuple[Path | None, str]:
    src = workdir / "sol.cpp"
    exe = workdir / "sol"
    src.write_text(code)
    proc = subprocess.run(
        ["g++", "-O2", "-std=c++17", "-o", str(exe), str(src)],
        capture_output=True, text=True, timeout=30,
    )
    if proc.returncode != 0:
        return None, proc.stderr
    return exe, ""


# ---------- 3. Run a single test ----------
def run_binary(exe: Path, stdin_data: str, time_limit: float) -> tuple[str, str]:
    """Returns (status, stdout). status in {'ok', 'tle', 'rte'}."""
    try:
        proc = subprocess.run(
            [str(exe)],
            input=stdin_data,
            capture_output=True,
            text=True,
            timeout=time_limit + 0.5,  # small slack
        )
    except subprocess.TimeoutExpired:
        return "tle", ""
    if proc.returncode != 0:
        return "rte", proc.stdout
    return "ok", proc.stdout


# ---------- 4. Output checking ----------
def normalize(s: str) -> list[list[str]]:
    """Strip trailing whitespace per line, drop trailing blank lines, split tokens."""
    lines = [ln.rstrip() for ln in s.replace("\r\n", "\n").split("\n")]
    while lines and lines[-1] == "":
        lines.pop()
    return [ln.split() for ln in lines]

def direct_match(expected: str, actual: str) -> bool:
    return normalize(expected) == normalize(actual)

def checker_match(checker_src: str, inp: str, ref_out: str, sub_out: str,
                  workdir: Path) -> bool:
    chk = workdir / "checker.py"
    chk.write_text(checker_src)
    inp_f = workdir / "in.txt"; inp_f.write_text(inp)
    ref_f = workdir / "ref.txt"; ref_f.write_text(ref_out)
    sub_f = workdir / "sub.txt"; sub_f.write_text(sub_out)
    try:
        proc = subprocess.run(
            ["python3", str(chk), str(inp_f), str(ref_f), str(sub_f)],
            capture_output=True, text=True, timeout=30,
        )
    except subprocess.TimeoutExpired:
        return False
    return proc.stdout.strip() == "1"


# ---------- 5. Evaluate one problem ----------
def evaluate_problem(row: dict, llm_response: str) -> dict:
    result = {
        "id": row["id"],
        "total": 0, "passed": 0, "accuracy": 0.0,
        "compile_ok": False, "error": None,
        "per_test": [],  # list of 'ok'/'wa'/'tle'/'rte'
    }

    code = extract_cpp(llm_response)
    if code is None:
        result["error"] = "no_code_extracted"
        return result

    tests = row.get("official_tests") or []
    if isinstance(tests, str):
        tests = ast.literal_eval(tests)
    if not tests:
        result["error"] = "no_tests"
        return result

    time_limit = float(row.get("time_limit") or 1.0)
    checker_src = row.get("generated_checker")

    with tempfile.TemporaryDirectory() as td:
        workdir = Path(td)
        exe, err = compile_cpp(code, workdir)
        if exe is None:
            result["error"] = f"compile_error: {err[:500]}"
            return result
        result["compile_ok"] = True

        for t in tests:
            result["total"] += 1
            status, out = run_binary(exe, t["input"], time_limit)
            if status == "tle":
                result["per_test"].append("tle"); continue
            if status == "rte":
                result["per_test"].append("rte"); continue

            if checker_src:
                ok = checker_match(checker_src, t["input"], t["output"], out, workdir)
            else:
                ok = direct_match(t["output"], out)

            if ok:
                result["passed"] += 1
                result["per_test"].append("ok")
            else:
                result["per_test"].append("wa")

    if result["total"]:
        result["accuracy"] = result["passed"] / result["total"]
    return result


# ---------- 6. Run across df in parallel ----------
def _worker(args):
    row_dict, llm_response = args
    return evaluate_problem(row_dict, llm_response)

def evaluate_dataframe(df, response_col="llm_response", n_workers=8):
    tasks = [(row, row[response_col]) for row in df.to_dict("records")]
    results = [None] * len(tasks)
    with ProcessPoolExecutor(max_workers=n_workers) as ex:
        futs = {ex.submit(_worker, t): i for i, t in enumerate(tasks)}
        for fut in as_completed(futs):
            results[futs[fut]] = fut.result()
    # attach to df
    df = df.copy()
    df["accuracy"]   = [r["accuracy"]   for r in results]
    df["passed"]     = [r["passed"]     for r in results]
    df["total"]      = [r["total"]      for r in results]
    df["compile_ok"] = [r["compile_ok"] for r in results]
    df["eval_error"] = [r["error"]      for r in results]
    df["per_test"]   = [r["per_test"]   for r in results]
    return df
````

````python
import subprocess
import tempfile
import os
import re
from typing import Optional, Tuple

def extract_code(completion: str) -> Tuple[Optional[str], Optional[str]]:
    match = re.search(r"```(python|cpp)\n(.*?)```", completion, re.DOTALL)
    if match:
        return match.group(1), match.group(2).strip()
    # If no markdown fences, treat as raw code (default python)
    if completion.strip():
        return "python", completion.strip()
    return None, None


def run_code(lang: str, code: str, test_input: str, timeout: int = 5) -> Optional[str]:
    """Run code with given input, return stdout or None on failure."""
    with tempfile.NamedTemporaryFile(
        mode='w',
        suffix='.py' if lang == 'python' else '.cpp',
        delete=False
    ) as tmp:
        tmp.write(code)
        tmp_path = tmp.name

    try:
        if lang == 'python':
            result = subprocess.run(
                ['python3', tmp_path],
                input=test_input,
                capture_output=True,
                text=True,
                timeout=timeout
            )
            return result.stdout.strip()

        else:  # cpp
            exe_path = tmp_path + '.out'
            compile_result = subprocess.run(
                ['g++', '-std=c++17', tmp_path, '-o', exe_path],
                capture_output=True,
                text=True,
                timeout=10
            )
            if compile_result.returncode != 0:
                print(f"  [Compile Error] {compile_result.stderr[:200]}")
                return None

            result = subprocess.run(
                [exe_path],
                input=test_input,
                capture_output=True,
                text=True,
                timeout=timeout
            )
            os.remove(exe_path)
            return result.stdout.strip()

    except subprocess.TimeoutExpired:
        print("  [TLE] Time limit exceeded")
        return None
    except Exception as e:
        print(f"  [Error] {e}")
        return None
    finally:
        if os.path.exists(tmp_path):
            os.remove(tmp_path)


def normalize_output(s: str) -> str:
    """Normalize line endings and strip trailing whitespace per line."""
    return "\n".join(line.rstrip() for line in s.replace('\r\n', '\n').strip().split('\n'))


def evaluate_solution(row, solution_code: str, verbose: bool = True) -> dict:
    """
    Evaluate a solution against all test cases in a dataframe row.

    Args:
        row:           A pandas Series (one row of your dataframe).
        solution_code: The solution as a string (raw code or markdown-fenced).
        verbose:       Print per-test results.

    Returns:
        dict with keys: passed, total, results
    """
    lang, code = extract_code(solution_code)
    if not code:
        print("Could not extract code.")
        return {"passed": 0, "total": 0, "results": []}

    # ── Collect all test cases ──────────────────────────────────────────────
    test_cases = []

    # Official tests
    official = list(row.get("official_tests", []))
    if isinstance(official, list):
        for t in official:
            if isinstance(t, dict) and "input" in t and "output" in t:
                # print(t['input'])
                test_cases.append({
                    "input":    t["input"].replace('\r\n', '\n'),
                    "expected": t["output"].replace('\r\n', '\n'),
                    "source":   "official"
                })

    # # Generated tests — stored as separate input/output lists or dicts
    # generated = row.get("generated_tests", [])
    # if isinstance(generated, list):
    #     for t in generated:
    #         if isinstance(t, dict) and "input" in t and "output" in t:
                
    #             test_cases.append({
    #                 "input":    str(t["input"]).replace('\r\n', '\n'),
    #                 "expected": str(t["output"]).replace('\r\n', '\n'),
    #                 "source":   "generated"
    #             })

    if not test_cases:
        print("No test cases found in row.")
        return {"passed": 0, "total": 0, "results": []}

    # ── Run each test ───────────────────────────────────────────────────────
    results = []
    passed  = 0

    for i, tc in enumerate(test_cases):
        actual = run_code(lang, code, tc["input"])
        ok     = actual is not None and \
                 normalize_output(actual) == normalize_output(tc["expected"])

        if ok:
            passed += 1

        results.append({
            "index":    i + 1,
            "source":   tc["source"],
            "passed":   ok,
            "expected": tc["expected"].strip(),
            "actual":   actual if actual is not None else "<no output>",
        })

        if verbose:
            status = "✅ PASS" if ok else "❌ FAIL"
            print(f"  Test {i+1:>2} [{tc['source']:9}] {status}")
            if not ok:
                print(f"    Expected : {tc['expected'].strip()[:120]}")
                print(f"    Got      : {str(actual)[:120]}")

    print(f"\n{'='*45}")
    print(f"  Result: {passed} / {len(test_cases)} test cases passed")
    print(f"{'='*45}")

    return {"passed": passed, "total": len(test_cases), "results": results}

solution = """
```python
import sys
input = sys.stdin.readline

t = int(input())
for _ in range(t):
    n = int(input())
    s = input().strip()
    score = 0
    for c in "ABCD":
        score += min(s.count(c), n)
    print(score)
```
"""
row = df_test_final.iloc[0]
# row is a pandas Series from your dataframe
result = evaluate_solution(row, solution, verbose=True)
print(result["passed"], "/", result["total"])
````

````python
def build_prompt(row) -> str:
    HEADER = (
        "You will be given a competitive programming problem. "
        "Please reason step by step about the solution in <think>...</think> tags, then provide a complete implementation in C++17.\n\n"
        "Your solution must read input from standard input (cin), write output to standard output (cout).\n"
        "Do not include any debug prints or additional output.\n\n"
        "Put your final solution within a single code block:\n"
        "```cpp\n<your code here>\n```\n\n"
    )

    # --- Problem section ---
    parts = [HEADER, "# Problem\n\n", row["description"].strip(), "\n\n"]

    # Time / memory limits
    parts.append(f"Execution time limit: {row['time_limit']} seconds\n")
    parts.append(f"Memory limit: {row['memory_limit']} MB\n\n")

    # Input format
    if row.get("input_format"):
        parts.append("## Input Format\n")
        parts.append(row["input_format"].strip())
        parts.append("\n\n")

    # Output format
    if row.get("output_format"):
        parts.append("## Output Format\n")
        parts.append(row["output_format"].strip())
        parts.append("\n\n")

    # Examples
    examples = row.get("examples")
    if examples.any():
        # examples may be a list of dicts or a stringified list — handle both
        if isinstance(examples, str):
            import ast
            examples = ast.literal_eval(examples)

        parts.append("## Examples\n")
        for ex in examples:
            parts.append("```input\n")
            parts.append(ex["input"].strip())
            parts.append("\n```\n")
            parts.append("```output\n")
            parts.append(ex["output"].strip())
            parts.append("\n```\n\n")

    # Note (optional)
    note = row.get("note", "").strip() if row.get("note") else ""
    if note and note.lower() != "none":
        parts.append("## Note\n")
        parts.append(note)
        parts.append("\n")

    return "".join(parts)
````
