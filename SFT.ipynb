````python
from unsloth import FastLanguageModel, is_bfloat16_supported
from unsloth.chat_templates import get_chat_template, train_on_responses_only
from datasets import Dataset, concatenate_datasets
from trl import SFTTrainer, SFTConfig

MODEL_NAME   = "unsloth/Qwen3-4B-unsloth-bnb-4bit"
MAX_SEQ_LEN  = 4096   # reasoning traces + cpp code need room

# ── Model ────────────────────────────────────────────────────────────
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name      = MODEL_NAME,
    max_seq_length  = MAX_SEQ_LEN,
    load_in_4bit    = True,
    full_finetuning = False,
)

model = FastLanguageModel.get_peft_model(
    model,
    r                        = 32,               # ↑ from 16; reasoning needs capacity
    target_modules           = ["q_proj","k_proj","v_proj","o_proj",
                                "gate_proj","up_proj","down_proj"],
    lora_alpha               = 32,               # alpha == r
    lora_dropout             = 0.05,             # tiny dropout curbs repetition
    bias                     = "none",
    use_gradient_checkpointing = "unsloth",
    random_state             = 3407,
    use_rslora               = True,             # more stable at higher r
)
````

````python
tokenizer
````

````python
SYSTEM_PROMPT = ("You're an expert competitive C++ programmer"
                 "Do all the reasoning before giving code output in "
                 "<think>..</think> tags. Just give the code in ```cpp ``` "
                 "after completing reasoning")
````

````python
full_ds = Dataset.from_parquet("final_ds.parquet")

def format_example(ex):
    msgs = [{"role": "system", "content": SYSTEM_PROMPT}] + list(ex["messages"])
    return {"text": tokenizer.apply_chat_template(
        msgs, tokenize=False, add_generation_prompt=False
    )}

full_ds = full_ds.map(format_example, num_proc=4,
                      remove_columns=[c for c in full_ds.column_names if c != "index"])

a_ds = full_ds.filter(lambda x: x["index"] == "A").shuffle(seed=42)
b_ds = full_ds.filter(lambda x: x["index"] == "B").shuffle(seed=42)
c_ds = full_ds.filter(lambda x: x["index"] == "C").shuffle(seed=42)

# Progressive curriculum: each stage keeps older data to prevent forgetting
stage1 = a_ds                                                   # ~7.6k
stage2 = concatenate_datasets([a_ds, b_ds]).shuffle(seed=42)    # ~12.8k
stage3 = concatenate_datasets([a_ds, b_ds, c_ds]).shuffle(seed=42)  # ~15k
````

````python
def make_trainer(ds, output_dir, epochs=1, lr=2e-4):
    trainer = SFTTrainer(
        model           = model,
        tokenizer       = tokenizer,
        train_dataset   = ds,
        args = SFTConfig(
            max_seq_length     = MAX_SEQ_LEN,
            dataset_text_field = "text",
            packing            = False,
            per_device_train_batch_size = 4,     # bump to 8 if VRAM allows
            gradient_accumulation_steps = 4,     # effective batch 16
            warmup_ratio                = 0.03,
            num_train_epochs            = epochs,
            learning_rate               = lr,
            logging_steps               = 10,
            save_steps                  = 60,
            save_total_limit            = 2,
            optim                       = "adamw_8bit",
            weight_decay                = 0.01,
            lr_scheduler_type           = "cosine",   # cosine > linear on small data
            seed                        = 3407,
            output_dir                  = output_dir,
            report_to                   = "none",
            bf16                        = is_bfloat16_supported(),
            fp16                        = not is_bfloat16_supported(),
            neftune_noise_alpha         = 5,          # regularizes, reduces loops
        ),
    )
    # CRITICAL: mask prompt tokens from the loss
    return train_on_responses_only(
        trainer,
        instruction_part = "<|im_start|>user\n",
        response_part    = "<|im_start|>assistant\n",
    )
````

````python
print("── Stage 1: A only (format + basic reasoning) ──")
make_trainer(stage1, "outputs/stage1_A", epochs=1, lr=2e-4).train()

print("── Stage 2: A + B (build up) ──")
make_trainer(stage2, "outputs/stage2_AB", epochs=1, lr=1e-4).train()

print("── Stage 3: A + B + C (full difficulty) ──")
make_trainer(stage3, "outputs/stage3_ABC", epochs=1, lr=5e-5).train()

model.save_pretrained("qwen3_4b_cf_lora")
tokenizer.save_pretrained("qwen3_4b_cf_lora")
````

````python
messages = [
    {"role" : "user", "content" : "Solve (x + 2)^2 = 0."}
]
text = tokenizer.apply_chat_template(
    messages,
    tokenize = False,
    add_generation_prompt = True, # Must add for generation
    enable_thinking = False, # Disable thinking
)

from transformers import TextStreamer
_ = model.generate(
    **tokenizer(text, return_tensors = "pt").to("cuda"),
    max_new_tokens = 450, # Increase for longer outputs!
    temperature = 0.7, top_p = 0.8, top_k = 20, # For non thinking
    streamer = TextStreamer(tokenizer, skip_prompt = False),
)
````
